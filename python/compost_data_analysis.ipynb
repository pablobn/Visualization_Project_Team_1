```python
# Import modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('combined_compost_measurements (2).csv')
df
```

```python
# Get the copy of original data set
df_copy = df.copy()
df_copy
```

```python
df_copy.isna().sum()
```

```python
# Convert the day into date-time
df_copy['Day'] = pd.to_datetime(df_copy['Day'])
df_copy
```

```python
# Filtered the dataset with essential columns
df_copy.columns
```

### How the internal temperature varies with out-side temperature

```python
columns = [
    'Day', 'Compost 1-Inside-Heating - w',
       'Compost 1-Lower-Moisture', 'Compost 1-Lower-Temperature',
       'Compost 1-Middle-Moisture', 'Compost 1-Middle-Temperature',
       'Compost 1-Upper-Moisture', 'Compost 1-Upper-Temperature',
       'Compost 2-Inside-Heating - w',
       'Compost 2-Lower-Moisture', 'Compost 2-Lower-Temperature',
       'Compost 2-Middle-Moisture', 'Compost 2-Middle-Temperature',
       'Compost 2-Upper-Moisture', 'Compost 2-Upper-Temperature',
       'Outside-Outside-Moisture', 'Outside-Outside-Temperature',
]
# Select Compost 1 relevant columns
compost1_cols = [
    'Day',
    'Compost 1-Inside-Heating - w',
    'Compost 1-Lower-Temperature',
    'Compost 1-Middle-Temperature',
    'Compost 1-Upper-Temperature',
    'Outside-Outside-Temperature'
]

df_c1 = df_copy[compost1_cols]
df_c1
```

```python
# We can convert this wide format to long format by using  melt()
df_c1_long = df_c1.melt(
    id_vars = "Day",
    var_name = "Temperature",
    value_name = "Values"
)
df_c1_long
```

```python
# Set columns for compost 2
compost2_cols = [
    'Day',
    'Compost 2-Inside-Heating - w',
    'Compost 2-Lower-Temperature',
    'Compost 2-Middle-Temperature',
    'Compost 2-Upper-Temperature',
    'Outside-Outside-Temperature'
]
df_c2 = df_copy[compost2_cols]
```

```python
# Convert this wide format into long format using melt()
df_c2_long = df_c2.melt(
    id_vars = "Day",
    var_name = "Temperature",
    value_name = "Values" 
)
df_c2_long
```

```python
# Visulaize all via lineplot
fig, axes = plt.subplots(2, 1, figsize=(14,10), sharex=True)

# Compost 1
sns.lineplot(data=df_c1_long, x='Day', y='Values', hue='Temperature', ax=axes[0])
axes[0].set_title("Compost 1 - Temperature Varies with Out-Side Temperature")

# Compost 2
sns.lineplot(data=df_c2_long, x='Day', y='Values', hue='Temperature', ax=axes[1])
axes[1].set_title("Compost 2 - Temperature Varies with Out-Side Temperature ")

plt.xticks(rotation=45)
plt.show()
```

# 📊 Correlation Analysis of Compost Layers

## 🔍 Objective
To evaluate how different compost layers respond to external environmental conditions, correlation analysis was performed between:

- Layer temperatures and outside temperature  
- Layer moisture and outside moisture  

This analysis helps identify the most stable and effective composting zone.

---

## 🌡️ Temperature Correlation with Outside Environment

### Compost 1
- **Lower Layer:** 0.44  
- **Middle Layer:** 0.20  
- **Upper Layer:** 0.07  

The upper layer shows the lowest correlation with outside temperature. However, this layer is directly exposed to environmental conditions, making it more prone to heat loss and fluctuations.

The **middle layer demonstrates relatively low correlation while remaining insulated**, indicating a stable thermal environment suitable for microbial activity.

---

### Compost 2
- **Lower Layer:** 0.65  
- **Middle Layer:** 0.52  
- **Upper Layer:** 0.44  

All layers in Compost 2 show higher correlation values compared to Compost 1, suggesting that the compost is more influenced by external temperature changes. This indicates weaker insulation and less efficient heat retention.

---

## 💧 Moisture Correlation with Outside Environment

### Compost 1
- **Lower Layer:** -0.49  
- **Middle Layer:** -0.57  
- **Upper Layer:** -0.29  

The middle layer shows the strongest resistance to external moisture variations, indicating better moisture stability.

---

### Compost 2
- **Lower Layer:** -0.52  
- **Middle Layer:** -0.44  
- **Upper Layer:** -0.54  

Moisture levels across all layers are similarly affected by outside conditions, suggesting a lack of internal stability.

---

## 🧠 Overall Interpretation

- The **upper layer** is highly exposed to environmental conditions and is less reliable for stable composting.  
- The **lower layer** is influenced by base conditions and shows moderate variability.  
- The **middle layer** remains the most stable in terms of both temperature and moisture.  

---

## 🏆 Final Conclusion

The **middle layer in Compost 1** provides the most favorable conditions for composting. It maintains a stable temperature and moisture level while being less affected by external environmental changes. This stability supports optimal microbial activity, making it the most effective layer for compost development.

In contrast, Compost 2 shows higher sensitivity to outside conditions across all layers, indicating less efficient composting performance.
